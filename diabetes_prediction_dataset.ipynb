```python
print("Gender values:")
print(df["gender"].unique())

print("\nSmoking history values:")
print(df["smoking_history"].unique())
```

```python
gender_mapping = {
    "Female": 0,
    "Male": 1,
    "Other": 2
}

df["gender"] = df["gender"].map(gender_mapping)
```

```python
smoking_mapping = {
    "never": 0,
    "current": 1,
    "former": 2,
    "ever": 3,
    "not current": 4,
    "No Info": 5
}

df["smoking_history"] = df["smoking_history"].map(
    smoking_mapping
)
```

```python
df.to_csv("diabetes_encoded.csv", index=False)

print("CSV file saved successfully!")
```
